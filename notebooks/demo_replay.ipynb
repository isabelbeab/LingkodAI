# LingkodAI demo: replay saved conversations

No GPU and no models: this plays back conversations that already ran, with
the same per-stage view as the live demos. It reads:

- sessions saved by `demo_chat.ipynb` and `demo_audio.ipynb`
  (`MyDrive/lingkodai_demo_sessions/` on Colab, `outputs/demo_sessions/`
  locally), with the spoken questions for audio sessions;
- golden-check results saved by `colab_golden_check.ipynb`
  (`MyDrive/lingkodai_golden_check/`) or written by `scripts/golden_check.py`
  (`outputs/golden_check/`);
- any folder from `scripts/run_conversation.py`.

The A/B notebook's runs are a different layout; that notebook has its own
listening step.

**Voice notice.** Synthesized audio here uses a cloned voice from a Philippine
Languages Database research speaker, under a CC-BY-NC, research-only licence.
It is not a real PRC official. Keep the audio and any saved sessions private.

## 1. Setup

On Colab, a CPU runtime is enough: this clones the repo (nothing to install)
and mounts Google Drive. Locally it uses your LingkodAI checkout
(`~/LingkodAI`, or set `LINGKODAI_DIR`).

```python
BRANCH = "demo-notebooks"
import os, subprocess, sys

IN_COLAB = "google.colab" in sys.modules
if IN_COLAB:
    REPO = "/content/LingkodAI"
    if not os.path.isdir(REPO):
        subprocess.run(["git", "clone", "-q", "--branch", BRANCH, "https://github.com/isabelbeab/LingkodAI.git", REPO], check=True)
    from google.colab import drive
    drive.mount("/content/drive")
    ROOTS = ["/content/drive/MyDrive/lingkodai_demo_sessions", "/content/drive/MyDrive/lingkodai_golden_check"]
else:
    REPO = os.path.abspath(os.path.expanduser(os.environ.get("LINGKODAI_DIR", "~/LingkodAI")))
    ROOTS = [f"{REPO}/outputs/demo_sessions", f"{REPO}/outputs/golden_check"]
if REPO not in sys.path:
    sys.path.insert(0, REPO)
from src import demo_ui
```

## 2. Find saved conversations

Newest first. Add any other folder to `ROOTS` and rerun.

```python
SESSIONS = []
for root in ROOTS:
    if os.path.isdir(root):
        SESSIONS += demo_ui.find_sessions(root)
    else:
        print("not found:", root)
for i, folder in enumerate(SESSIONS):
    manifest, turns = demo_ui.load_session(folder)
    langs = sorted({t.get("final_lang") for t in turns if t.get("final_lang")})
    print(f"{i:>3}  {folder}  {len(turns)} turn(s), {', '.join(langs)}")
```

## 3. Play one

`PICK` is a number from the list above, or a folder path.

```python
PICK = 0
demo_ui.show_session(SESSIONS[PICK] if isinstance(PICK, int) else PICK)
```
